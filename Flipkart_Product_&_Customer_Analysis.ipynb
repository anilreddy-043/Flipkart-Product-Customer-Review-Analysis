# Flipkart Product & Customer Review Analysis

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

path = r"C:\Users\anilr\OneDrive\Desktop\Flipkart Product & Customer Review Analysis\Flipkart_Product_cleaned_data_set.csv"
df = pd.read_csv(path, encoding="latin1")

df.head()
```

```python
sns.histplot(data=df, x='Rate', bins=10)
plt.title('Distribution of Product Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Products')
plt.show()
```

```python
sns.histplot(data=df, x='Price', bins=30)
plt.title('Distribution of Product Prices')
plt.xlabel('Price')
plt.ylabel('Number of Products')
plt.show()
```

```python
sns.barplot(data=df, x='Category', y='Rate', estimator='mean')
plt.xticks(rotation=45)
plt.title('Average Rating by Category')
plt.show()
```

```python
sns.scatterplot(data=df, x='Price', y='Rate')
plt.title('Price vs Rating')
plt.xlabel('Price')
plt.ylabel('Rating')
plt.show()
```

```python
sns.countplot(data=df, x='Category')
plt.xticks(rotation=45)
plt.title('Number of Products by Category')
plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
top_products = df.sort_values('Rate', ascending=False).head(10)

sns.barplot(data=top_products, x='Rate', y='Product_Name')
plt.title('Top 10 Products by Rating')
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='Price',
    y='Rate',
    hue='Category'
)

plt.title('Price vs Rating by Category')
plt.show()
```

```python
sns.boxplot(data=df, x='Category', y='Price')
plt.xticks(rotation=45)
plt.title('Price Distribution by Category')
plt.show()
```
