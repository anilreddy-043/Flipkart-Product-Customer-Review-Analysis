```python
import pandas as pd

path = r"C:\Users\anilr\OneDrive\Desktop\Flipkart Product & Customer Review Analysis\Dataset.csv"

df = pd.read_csv(path, encoding="latin1")

df.head()
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
df['Rate'] = pd.to_numeric(df['Rate'], errors='coerce').astype(float)
```

```python
df['Price'] = pd.to_numeric(
    df['Price'].str.replace(r'[^\d]', '', regex=True),
    errors='coerce'
)
```

```python
df.info()
```

```python
df.dtypes
```

```python
df['Product_name'] = df['Product_name'].fillna(df['Product_name'].mode()[0])
df['Product_name'].isna().sum()
```

```python
df['Review'] = df['Review'].fillna(df['Review'].mode()[0])
df['Review'].isna().sum()
```

```python
df['Summary'] = df['Summary'].fillna(df['Product_name'].mode()[0])
df['Summary'].isna().sum()
```

```python
df['Price'] = df['Price'].fillna(df['Price'].mean())
df['Price'].isna().sum()
```

```python
df['Rate'] = df['Rate'].fillna(df['Rate'].mean())
df['Rate'].isna().sum()
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```
